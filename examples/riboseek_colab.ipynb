# RiboSeek — RNA 3D structure search in Colab

Runs the released `riboseek` package: install, download the 15,391-chain encoded database once, upload a structure (PDB or mmCIF), get a ranked list of similar RNA chains with E-values.

Runtime: CPU is enough (a query takes well under a second after the one-off index build).

```python
!pip -q install riboseek
!riboseek info
```

```python
!riboseek download-db
```

## Upload a query structure
Any `.pdb`, `.cif`, `.mmcif` (optionally `.gz`). The longest RNA chain is used unless you pass `--chain`.

```python
from google.colab import files
up = files.upload()
query = next(iter(up))
print('query:', query)
```

```python
!riboseek search "$query" --top-n 10
```

## Optional: geometry-only alphabet (RS-20), a specific chain, or folding modified nucleotides onto their parent base

```python
!riboseek search "$query" --top-n 10 --alphabet sa20
#!riboseek search "$query" --chain A --keep-modified
```

## Python API

```python
from riboseek import Searcher
s = Searcher.from_pretrained()          # RS-80 with E-values
hits = s.search(query, top_n=5)
for h in hits:
    print(f"{h['chain']:>12}  combined {h['combined_score']:+.3f}  E = {h['evalue']:.2e}  len {h['length']}")
```

## Batch mode
Upload several structures into a folder and give the folder to `riboseek search`; add `--tsv hits.tsv` to save all hits.

```python
!mkdir -p queries && cp "$query" queries/
!riboseek search queries --top-n 5 --tsv hits.tsv && head hits.tsv
```
